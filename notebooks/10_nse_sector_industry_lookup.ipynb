```python
# Merge NSE constituent CSV files

import pandas as pd
import glob
import os

files = glob.glob(
    "/lakehouse/default/Files/data/raw/nse_sector_lookup/*.csv"
)

all_data = []

for file in files:

    try:

        df = pd.read_csv(file)

        # index name from file name
        index_name = (
            os.path.basename(file)
            .replace(".csv", "")
        )

        df["index_name"] = index_name

        df = df[[
            "Symbol",
            "ISIN Code",
            "Industry",
            "index_name"
        ]]

        df = df.rename(columns={
            "Symbol": "symbol",
            "ISIN Code": "isin",
            "Industry": "industry"
        })

        all_data.append(df)

    except Exception as e:

        print(
            f"ERROR -> {os.path.basename(file)} : {e}"
        )

lookup_df = pd.concat(
    all_data,
    ignore_index=True
)

# remove exact duplicates
lookup_df = lookup_df.drop_duplicates()

save_path = (
    "/lakehouse/default/Files/data/processed/"
    "securities/sector_industry_lookup_nse.csv"
)

lookup_df.to_csv(
    save_path,
    index=False
)

print("SUCCESS")
print("Files:", len(files))
print("Rows:", len(lookup_df))
print(
    "Unique Symbols:",
    lookup_df["symbol"].nunique()
)
print(
    "Unique Industries:",
    lookup_df["industry"].nunique()
)
print("Saved:", save_path)
```
