## **NSE Trade Historical**

```python
import pandas as pd
import yfinance as yf
import os
from datetime import date


# Read NSE security file
df_sec = pd.read_csv(
    "/lakehouse/default/Files/data/raw/nse_sec/nse_security.csv"
)

# Get all unique NSE symbols
symbols = (
    df_sec["SYMBOL"]
    .dropna()
    .astype(str)
    .str.strip()
    .unique()
)

START_DATE = "2001-01-01"
END_DATE = date.today().isoformat()

SAVE_DIR = "/lakehouse/default/Files/data/raw/nse_trade_hist"


for symbol in symbols:
    try:
        ticker = f"{symbol}.NS"

        print(f"Downloading {ticker}")

        df = yf.download(
            ticker,
            start=START_DATE,
            end=END_DATE,
            interval="1d",
            progress=False
        )

        # Fix extra yfinance ticker header
        df.columns = df.columns.get_level_values(0)

        if df.empty:
            print(f"{symbol}: no data")
            continue

        df = df.reset_index()
        df.insert(0, "Symbol", symbol)

        file_path = os.path.join(
            SAVE_DIR,
            f"{symbol}.csv"
        )

        df.to_csv(file_path, index=False)

        print(f"{symbol}: saved")

    except Exception as e:
        print(f"{symbol}: ERROR -> {e}")

print("NSE historical download completed")
```

```python
import pandas as pd
import os
from datetime import datetime, timedelta
from nselib import capital_market


SAVE_DIR = "/lakehouse/default/Files/data/raw/nse_block_hist"
os.makedirs(SAVE_DIR, exist_ok=True)

all_data = []

start_date = datetime(2001, 1, 1)
end_date = datetime.today()


current_start = start_date

while current_start <= end_date:

    current_end = current_start + timedelta(days=6)

    if current_end > end_date:
        current_end = end_date

    from_dt = current_start.strftime("%d-%m-%Y")
    to_dt = current_end.strftime("%d-%m-%Y")

    print(f"Fetching {from_dt} → {to_dt}")

    try:
        data = capital_market.block_deals_data(
            from_date=from_dt,
            to_date=to_dt
        )

        df_chunk = pd.DataFrame(data)

        print(f"Rows: {len(df_chunk)}")

        all_data.append(df_chunk)

    except Exception as e:
        print(f"ERROR -> {e}")

    current_start = current_end + timedelta(days=1)


# Combine all chunks
df = pd.concat(all_data, ignore_index=True)

print("Total rows:", len(df))


today = datetime.today().strftime("%d-%m-%Y")

file_name = f"blockdeals_{today}.csv"

file_path = os.path.join(
    SAVE_DIR,
    file_name
)

df.to_csv(file_path, index=False)

print("Saved:", file_path)
```

```python
# correct version
import pandas as pd
import requests
import io
import os
import time
from datetime import datetime
from calendar import monthrange


SAVE_DIR = "/lakehouse/default/Files/data/raw/nse_bulk_hist"
os.makedirs(SAVE_DIR, exist_ok=True)

session = requests.Session()

headers = {
    "User-Agent": "Mozilla/5.0",
    "Accept": "text/csv,*/*",
    "Accept-Encoding": "gzip, deflate",   # disable brotli
    "Referer": "https://www.nseindia.com/"
}

# Get cookies
session.get(
    "https://www.nseindia.com",
    headers=headers
)

all_data = []

current_year = datetime.today().year
current_month = datetime.today().month


for year in range(2001, current_year + 1):

    print(f"\nProcessing {year}")

    for month in range(1, 13):

        if year == current_year and month > current_month:
            break

        from_dt = f"01-{month:02d}-{year}"

        last_day = monthrange(year, month)[1]

        to_dt = f"{last_day:02d}-{month:02d}-{year}"

        if year == current_year and month == current_month:
            to_dt = datetime.today().strftime("%d-%m-%Y")

        print(f"Fetching {from_dt} → {to_dt}")

        try:

            url = (
                "https://www.nseindia.com/api/"
                "historicalOR/bulk-block-short-deals"
                f"?optionType=bulk_deals"
                f"&from={from_dt}"
                f"&to={to_dt}"
                f"&csv=true"
            )

            response = session.get(
                url,
                headers=headers,
                timeout=30
            )

            if response.status_code != 200:
                print(f"Failed: {response.status_code}")
                continue

            df_month = pd.read_csv(
                io.StringIO(response.text)
            )

            # clean weird column names
            df_month.columns = (
                df_month.columns
                .str.replace("ï»¿", "", regex=False)
                .str.strip()
            )

            print(f"Rows: {len(df_month)}")

            if not df_month.empty:
                all_data.append(df_month)

            time.sleep(0.5)

        except Exception as e:
            print(f"ERROR -> {e}")


# Combine all months
df_final = pd.concat(
    all_data,
    ignore_index=True
)

# Remove duplicates
df_final = df_final.drop_duplicates()

print("Final rows:", len(df_final))

today = datetime.today().strftime("%d-%m-%Y")

file_path = os.path.join(
    SAVE_DIR,
    f"bulkdeals_{today}.csv"
)

df_final.to_csv(
    file_path,
    index=False
)

print("Saved:", file_path)
```

## **BSE Block Historical**

```python
import pandas as pd
import glob
import os


folder_path = "/lakehouse/default/Files/data/raw/bse_block_hist"

files = glob.glob(
    os.path.join(folder_path, "*.csv")
)

print("Files found:", len(files))

all_data = []

for file in files:

    print("Reading:", os.path.basename(file))

    try:
        df = pd.read_csv(
            file,
            on_bad_lines="skip",
            low_memory=False
        )

        df.columns = df.columns.str.strip()

        print("Rows:", len(df))

        all_data.append(df)

    except Exception as e:
        print(f"ERROR -> {e}")


# Combine all files
df_final = pd.concat(
    all_data,
    ignore_index=True
)

# Remove duplicates
df_final = df_final.drop_duplicates()

# Convert Deal Date
df_final["Deal Date"] = pd.to_datetime(
    df_final["Deal Date"],
    dayfirst=True,
    errors="coerce"
)

# Sort by date
df_final = df_final.sort_values(
    "Deal Date"
)

# Convert back to readable format
df_final["Deal Date"] = (
    df_final["Deal Date"]
    .dt.strftime("%d-%m-%Y")
)

print("Final rows:", len(df_final))

file_path = os.path.join(
    folder_path,
    "blockdeals_01-06-2026.csv"
)

df_final.to_csv(
    file_path,
    index=False
)

print("Saved:", file_path)
```
