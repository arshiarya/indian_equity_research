```python
# Welcome to your new notebook
# Type here in the cell editor to add code!
bulk_df = spark.read.option(
    "header",
    "true"
).csv(
    "Files/data/processed/deals/bulk_processed.csv",
    inferSchema=True
)

print("Rows:", bulk_df.count())

bulk_df.printSchema()

display(
    bulk_df.limit(10)
)
```

```python
bulk_df = spark.read.option(
    "header",
    "true"
).csv(
    "Files/data/processed/deals/bulk_processed.csv",
    inferSchema=True
)

bulk_df.write.mode(
    "overwrite"
).format(
    "delta"
).saveAsTable(
    "bulk_daily"
)

print("SUCCESS")
print("Rows:", bulk_df.count())

display(
    bulk_df.limit(10)
)
```
